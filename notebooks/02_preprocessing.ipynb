# Part 2: Data Preprocessing & Train/Test Split
## Green-Synthesized ZnO Nanoparticle Research Project

This notebook handles data cleaning, feature selection, and leakage-free train/test splitting.

```python
import pandas as pd
from sklearn.model_selection import train_test_split

# Load dataset
df = pd.read_excel('../data/raw/zn_particle.xlsx')

# Define Features (X) for RQ1: Predict SEM Particle Size before characterization
feature_cols = ['pH', 'Temperature', 'Time', 'Zn_Concentration', 'Plant_Extract']
target_col = 'SEM_Size'

X = df[feature_cols]
y = df[target_col]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print(f"X_train shape: {X_train.shape}, X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}, y_test shape: {y_test.shape}")

# Save processed split datasets
X_train.to_csv('../data/processed/X_train.csv', index=False)
X_test.to_csv('../data/processed/X_test.csv', index=False)
y_train.to_csv('../data/processed/y_train.csv', index=False)
y_test.to_csv('../data/processed/y_test.csv', index=False)
print("Processed data splits saved successfully!")
```
